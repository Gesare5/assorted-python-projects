```python
import re


regex_pattern = r'([a-z]+)(_([a-z]+))+'  # regex pattern for finding snake_case variable names

sample_name = 'apple_type_juice'
sample_word = 'tree_tomato'

def is_snake_case(word):
    ''' Find string without the specified pattern '''
    replacement = re.search(regex_pattern,word)
    print(replacement)
    return bool(replacement)

print(is_snake_case(sample_word))
print(is_snake_case(sample_name))
```
