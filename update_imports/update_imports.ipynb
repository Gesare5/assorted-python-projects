```python
# Automation script to find and update all necessary imports

import os
import re


# regex_pattern = r'(\.{2}\/)+'    # Grab lines with ../ pattern or more
regex_pattern = r'(\.{2}\/){2,}'    # Grab lines with ../../ pattern or more

directory_name = './sample_folder'

def find_and_replace(line):
    ''' Returns string without the specified pattern '''
    replacement = re.sub(pattern=regex_pattern, repl='', string=line)
    return replacement

def read_file(name):
    with open(name, 'r' ) as in_file:  
        content = in_file.readlines()
    return content

def write_file(name, edited_content):
    with open(name, 'a') as out_file:
        out_file.writelines(edited_content)

def get_file_names(directory_name):
    ''' Get list of file names in a directory '''
    file_names = []
    for root, dirs, files in os.walk(directory_name, topdown=True):
        for name in files:
            name = os.path.join(root, name)
            file_names.append(name)        
    return file_names        



def main():
    file_names = get_file_names(directory_name)
    for name in file_names:
        edited_file_content = []
        # Read file and retrieve lines 
        content = read_file(name)
        # Get first 100 lines in file
        lines = content[0:100]
        rest = content[100:]

        # Remove file in order to replace it
        os.remove(name)

        for line in lines:
            # Make replacements for the first 100 lines
            edited_file_content.append(find_and_replace(line))
        edited_file_content.extend(rest)  

        write_file(name, edited_file_content)  
    

main()
```
